# Constrained research optimization

## Purpose and scientific status
Explore conversion, selectivity, throughput, and heat-duty tradeoffs within declared synthetic bounds.

## Inputs and units
Temperature: K. Residence time: s. Objectives and constraint keys are explicit in the evaluator.

## Execution and printed results
The notebook prints candidate parameters, objective metrics, status, and the mandatory safety statement.

## Validation and limitations
Optimization outputs are research hypotheses. They are not operating recommendations and require experimental validation plus process-safety review.

```python
import sys
from pathlib import Path
root=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd(); sys.path.insert(0,str(root/'src'))
from chemical_reactor_twin import ReactionNetwork, simulate_cstr
from chemical_reactor_twin.optimization import optimize_operating_conditions
```

```python
def evaluator(p):
 net=ReactionNetwork(['A','B','C'],[[-1,1,0],[-1,0,1]],[20,2],[12000,15000],[[1,0,0],[1,0,0]],[-20000,-30000])
 r=simulate_cstr(net,[1000,0,0],p['temperature_K'],p['residence_time_s'],ua_over_volume=25000,coolant_temperature=330)
 A,B,C=r.concentration[-1]; conv=1-A/1000; sel=B/max(B+C,1e-12)
 return {'conversion':float(conv),'selectivity':float(sel),'throughput':float(1/p['residence_time_s']),'heat_duty':float(abs(r.temperature[-1]-330)/10),'temperature_K':float(r.temperature[-1])}
result=optimize_operating_conditions(evaluator,{'temperature_K':(310,350),'residence_time_s':(10,120)},{'temperature_K':(1,345),'conversion':(0.2,1.0)})
print(result)
print('Safety statement:',result['status'])
```
