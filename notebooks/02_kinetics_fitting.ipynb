# Kinetics fitting and identifiability

Built with [Vicena](https://vicena.ai). Fit synthetic Arrhenius parameters. Do not transfer these values to real chemistry.

```python
import sys
from pathlib import Path
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd(); sys.path.insert(0,str(root/'src'))
import numpy as np
from chemical_reactor_twin import ReactionNetwork, simulate_batch, fit_arrhenius_parameters, identifiability_report
```

```python
truth=ReactionNetwork(['A','B'],[[-1,1]],[12.0],[11000.0],[[1,0]],[-10000.0])
experiments=[]
for T in [315.,335.,355.]:
 r=simulate_batch(truth,[1000,0],T,80,ua=1e9,coolant_temperature=T,n_points=41)
 experiments.append({'c0':[1000,0],'temperature_K':T,'time_s':r.coordinate,'species_index':0,'observed_concentration':r.concentration[:,0],'sigma':2.0})
fitnet=ReactionNetwork(['A','B'],[[-1,1]],[5.0],[8000.0],[[1,0]],[-10000.0])
fit=fit_arrhenius_parameters(fitnet,experiments)
ident=identifiability_report(fit['jacobian'])
print('fitted A',fit['A'],'true',12.0)
print('fitted Ea J/mol',fit['Ea_J_mol'],'true',11000.0)
print('cost',fit['cost']); print(ident)
```
